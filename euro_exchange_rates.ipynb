```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import lineStyles
```

```python
df=pd.read_csv('euro-daily-hist_1999_2022.csv')
df
```

```python
df=df.iloc[:,[0,1,4,-2]]
df.columns=['Date','Australian dollar','Canadian dollar','US dollar']
df
```

```python
df=pd.melt(df,id_vars='Date',value_vars=['Australian dollar','Canadian dollar','US dollar'],value_name='Euro rate',var_name='Currency')
df['Date']=pd.to_datetime(df['Date'])
df=df[df['Date']>='2022-12-01'].reset_index(drop=True)
df['Euro rate']=pd.to_numeric(df['Euro rate'])
print(f'Currency:{df.Currency.unique()}\n')
print(df)
```

```python
usd=df[df['Currency']=='US dollar']
sns.lineplot(x='Date',y='Euro rate',data=usd)
```

```python
fig=plt.subplots(figsize=(12,6))
sns.lineplot(x='Date',y='Euro rate',data=usd).set(title='Euro-US rate',xlabel='Date',ylabel='Euro')
```

```python
fig = plt.subplots(figsize=(12, 6))
sns.lineplot(x='Date', y='Euro rate', data=usd,linestyle='dotted',color='red',linewidth=5).set(title='Euro-US rate', xlabel='Date',ylabel='Euro')
```

```python
fig = plt.subplots(figsize=(12, 6))
sns.lineplot(x='Date', y='Euro rate', data=usd,marker='*',markerfacecolor='limegreen',markersize=5).set(title='Euro-US rate', xlabel='Date',ylabel='Euro')
```
